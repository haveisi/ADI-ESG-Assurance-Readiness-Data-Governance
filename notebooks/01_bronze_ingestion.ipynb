```python
%python
path = "/Volumes/esg_assurance/adi_assurance/bronze_supplier_procurement/raw_supplier_procurement.csv"

procurement_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(path)
)

display(procurement_df)
```

```python
%python
procurement_bronze = (
    procurement_df
    .withColumn("_ingested_at", F.current_timestamp())
    .withColumn("_source_file", F.lit(path))
    .withColumn("_source_system", F.lit("synthetic_procurement_erp"))
    .withColumn("_data_layer", F.lit("bronze"))
)
```
